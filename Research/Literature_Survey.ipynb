## Literature Survey for Project : Text-Based Data Extraction Pipeline for ML Ready Datasets.


### What is the point ?
Niche topics do not have ample open-source datasets for a model to be trained on. It takes a lot of time manually searching for data on the web.

**The project aims to connect these two things together =
Data-Scraping + Data Pre-Processing
In one single open-source pipeline.
Making all building blocks of this project to be open-source & free to use it the crux of this project**



### Already Existing Tools & Their Limitations

1. FireCrawl
  - Has url discovery constraints
  - High credit usage costs at scales
2. ScrapeGraphAI
  - steep technical setup curve
  - High credit usage costs at scales
3. LLM Scraper
  -  high operating costs
  - output hallucination risks

No tools right now can also label the extracted data.

### What the project is concerned with :
- **Scraping** the web for relevant data related to a specific topic. -> Web scrapers.
- **Storing, Cleaning & Sorting** the data into Datasets that can be directly used for Model Training.
- **Labelling the Extracted Data** = Ml Datasets often require labels.
- **Scalability Issues** = Extracting huge amounts of data requires significant processing time and compute.
- **Privacy & Security**

  (ref : https://www.semanticscholar.org/paper/e070db6ecbe0113b5318c146a32992e966c20a62)
